```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0, 1, 100)
```

```python
def sin_func(val):
    return np.sin(val)

def cos_func(val):
    return np.cos(val)
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2)

ax1.plot(x, sin_func(x))
ax1.set_title("sin(x)")
ax1.set_xlabel("x")

ax2.plot(x, cos_func(x))
ax2.set_title("cos(x)")
ax2.set_xlabel("x")

plt.savefig("session8.pdf")

plt.show()
```
